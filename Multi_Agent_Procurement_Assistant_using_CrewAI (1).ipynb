```python
data_agent = Agent(
    role="Product Data Collector",

    goal="""
    Collect laptop information from the dataset and web search results.
    """,

    backstory="""
    You are responsible for gathering structured product information
    including price, RAM, CPU, storage, GPU and operating system.
    """,

    llm=llm,

    verbose=True
)
```

```python
data_task = Task(
    description="""
Collect detailed specifications for every laptop returned by the Search Agent.

Include:

- Company
- Product
- Price
- CPU
- RAM
- Storage
- GPU
- Weight
- Operating System
""",

    expected_output="Structured laptop information.",

    agent=data_agent,

    context=[search_task]
)
```

```python
scraping_agent = Agent(
    role="Website Scraping Specialist",

    goal="""
    Read product pages and extract detailed specifications.
    """,

    backstory="""
    You specialize in extracting useful information from
    laptop product pages.
    """,

    llm=llm,

    tools=[search_tool],

    verbose=True
)
```

```python
scraping_task = Task(
    description="""
Extract detailed specifications from product webpages.

Include:

- CPU
- RAM
- Storage
- GPU
- Battery
- Weight
- Display
""",

    expected_output="Detailed specifications.",

    agent=scraping_agent,

    context=[data_task]
)
```

```python
comparison_agent = Agent(
    role="Laptop Comparison Expert",

    goal="""
    Compare laptops objectively.
    """,

    backstory="""
    You compare products according to
    price,
    specifications,
    performance,
    battery,
    portability,
    and value.
    """,

    llm=llm,

    verbose=True
)
```

```python
comparison_task = Task(
    description="""
Compare all collected laptops.

Create a comparison table including:

- Price
- CPU
- RAM
- Storage
- GPU
- Weight
- Battery Life

Explain the advantages and disadvantages of each laptop.

Recommend the best value for money.
""",

    expected_output="Laptop comparison table with recommendation.",

    agent=comparison_agent,

    context=[
        data_task,
        scraping_task
    ]
)
```

```python
ranking_agent = Agent(
    role="Procurement Decision Expert",

    goal="""
    Rank laptops from best to worst.
    """,

    backstory="""
    You evaluate products based on
    value for money,
    specifications,
    business suitability,
    and performance.
    """,

    llm=llm,

    verbose=True
)
```

```python
ranking_task = Task(
    description="""
Rank the laptops based on:

- Price
- CPU
- RAM
- Storage
- Portability
- Battery Life

Return the Top 5 products from best to worst.
""",

    expected_output="Ranked list of laptops.",

    agent=ranking_agent,

    context=[comparison_task]
)
```

```python
report_agent = Agent(
    role="Procurement Report Writer",

    goal="""
    Generate a professional procurement report based on the analysis.
    """,

    backstory="""
    You are a procurement consultant who prepares professional reports
    for companies after comparing products and analyzing specifications.
    """,

    llm=llm,

    verbose=True
)
```

```python
report_task = Task(
    description="""
Generate a professional HTML procurement report.

The report must include:

1. Executive Summary

2. Company Requirements

3. Product Comparison Table

4. Final Ranking

5. Procurement Recommendation

6. Conclusion
""",

    expected_output="Professional HTML report.",

    agent=report_agent,

    context=[ranking_task]
)
```

```python
crew = Crew(
    agents=[
        search_agent,
        data_agent,
        scraping_agent,
        comparison_agent,
        ranking_agent,
        report_agent
    ],

    tasks=[
        search_task,
        data_task,
        scraping_task,
        comparison_task,
        ranking_task,
        report_task
    ],

    process=Process.sequential,

    verbose=True
)
```

```python
import nest_asyncio

nest_asyncio.apply()

result = await crew.kickoff_async()

print(result)
```

```python
import requests
from bs4 import BeautifulSoup

def scrape_product(url):
    headers = {
        "User-Agent": "Mozilla/5.0"
    }

    response = requests.get(url, headers=headers)

    soup = BeautifulSoup(response.text, "html.parser")

    text = soup.get_text(" ", strip=True)

    return text[:5000]
```

```python
from crewai.tools import tool

@tool("Website Scraper")
def website_scraper(url: str) -> str:
    """
    Scrape webpage content.
    """
    return scrape_product(url)
```

```python
scraping_agent = Agent(
    role="Website Scraper",

    goal="""
    Extract laptop specifications from product webpages.
    """,

    backstory="""
    You are an expert web scraping specialist.
    Your responsibility is to extract accurate product information
    such as CPU, RAM, storage, GPU, battery life, weight,
    operating system, and price from laptop webpages.
    """,

    tools=[website_scraper],

    llm=llm,

    verbose=True
)
```

```python
scraping_task = Task(
    description="""
Use the Website Scraper tool to extract detailed information
from each laptop product webpage.

For each product extract:

- Product Name
- Company
- CPU
- RAM
- Storage
- GPU
- Weight
- Battery Life
- Operating System
- Price

Return the results as a structured table.
""",

    expected_output="""
A structured table containing the specifications of all laptops.
""",

    agent=scraping_agent,

    context=[search_task]
)
```
